**¿Cómo identificarías valores faltantes en un DataFrame de Pandas?**

`df.isna().sum()`: contar cuántos valores nulos (vacíos o faltantes) existen en cada columna de una tabla



```python
import pandas as pd

df = pd.DataFrame({
    "cliente": ["Ana", "Luis", None, "Pedro", "María"],
    "sucursal": ["Norte", None, "Sur", "Centro", None],
    "ventas": [1000, 2000, None, 1500, None]
})
print(df)
print(df.isna().sum())
```

**¿Qué hacer después de encontrar valores nulos?**

```python
# Eliminar filas que tengan algún valor nulo
df_limpio = df.dropna()

print(df_limpio)
```

```python
# Reemplazar valores nulos por cero
df["ventas"] = df["ventas"].fillna(0)

print(df)
```

```python
# Reemplazar valores nulos por el promedio
promedio = df["ventas"].mean()
df["ventas"] = df["ventas"].fillna(promedio)

print(df)
```

```python
df.isna().mean() * 100
```

**¿Qué significa `df.drop_duplicates()`?**

Sirve para eliminar filas duplicadas de un DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    "cliente": ["Ana", "Luis", "Ana", "Pedro", "Luis"],
    "producto": ["Laptop", "Mouse", "Laptop", "Monitor", "Mouse"],
    "ventas": [1500, 500, 1500, 2000, 500]
})
print(df)
sin_duplicados = df.drop_duplicates()
print(" ============================== ")
print(sin_duplicados)
```

**Formas de utilizar `drop_duplicates()`**


**`duplicated() `**→ identifica los duplicados.

**`duplicated().sum()`** → cuenta los duplicados.

**`drop_duplicates()`** → elimina duplicados.

**`drop_duplicates(subset=["cliente"])`** → elimina duplicados basándose en la columna cliente.

```python
# Eliminar filas completamente duplicadas
df.drop_duplicates()
```

```python
# Eliminar duplicados según una columna
df.drop_duplicates(subset="cliente")
```

```python
# Conservar el último registro
df.drop_duplicates(subset="cliente", keep="last")
```

```python
# Eliminar todas las apariciones duplicadas
df.drop_duplicates(keep=False)
```

**¿Qué diferencia hay entre groupby y pivot_table?**

`groupby() `: agrupa registros por una o varias categorías y permite calcular sumas, promedios, cantidades, máximos, etc.

`pivot_table()`: crea tablas dinámicas, similares a las de Excel, organizando los resultados en filas y columnas.

```python
import pandas as pd

df = pd.DataFrame({
    "sucursal": ["Norte", "Sur", "Norte", "Sur", "Centro", "Centro"],
    "producto": ["Laptop", "Laptop", "Mouse", "Mouse", "Laptop", "Mouse"],
    "ventas": [1000, 2000, 500, 800, 1500, 300]
})
print(df)
```

```python
# groupby() significa agrupar por
df.groupby("sucursal")["ventas"].sum()
```

```python
# Agrupar por dos columnas
df.groupby(["sucursal", "producto"])["ventas"].sum()
```

```python
# pivot_table() significa tabla dinámica
df.pivot_table(
    index = "sucursal",
    columns = "producto",
    values = "ventas",
    aggfunc = "sum"
)
```

**¿Cómo unir múltiples archivos CSV?**

`glob.glob("ventas/*.csv")`	: Todos los CSV dentro de ventas
`glob.glob("ventas/ventas_*.csv")`:	CSV que empiezan con ventas_
`glob.glob("datos/*.xlsx")`: Todos los Excel .xlsx de datos
`glob.glob("datos/**/*.csv", recursive=True)	`: CSV en datos y sus subcarpetas

```python
# EJERCICIO PRACTICO: CONSOLIDAR VENTAS Y CALCULAR TOTALES

#Importar librerias
import pandas as pd       # para trabajar los datos con las tablas
import glob               # buscar archivos utilizando patrones
import os                 # sistema operativo


# 1. Buscar los archivos CSV
# "ventas/*.csv" (patrón búsqueda): Le indica dónde y qué buscar "ventas/*" es la carpeta,
# el asterisco * es un comodín que significa "cualquier nombre"

# glob.glob(.) (buscador): función de librería glob que rastrea en computadora
# todos los archivos que coincidan con el patrón anterior y te devuelve una lista con sus rutas.

# sorted(...) (El ordenador): ordena alfabéticamente de la A a la Z
archivos = sorted(glob.glob("ventas/*.csv"))


# 2. Comprobar que encontramos archivos
if not archivos:
  print("No se encontraron archivos CSV en la carpeta")
else:
  # Leer los archivos
  lista_df = [pd.read_csv(a) for a in archivos]

  #unir archivos
  df_unido = pd.concat(lista_df, ignore_index=True)

  #Agrupar ventas por sucursal
  df.groupby("sucursal")["ventas"].sum()

  # crear carpeta de resultados
  os.makedirs("resultados", exist_ok=True)

  # Guardar el archivo
  df_unido.to_csv("resultados/ventas_totales.csv", index=False)
```
